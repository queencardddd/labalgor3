```python
#Лабораторная работа 3.
#Александрова Екатерина
#ЦИБ-251
'''
Алгоритм линейного поиска последовательно просматривает элементы массива слева направо
и сравнивает каждый элемент с искомым значением target.
Если элемент найден, функция возвращает индекс его первого вхождения.
Если после проверки всех элементов совпадение не найдено, функция возвращает -1.
Сложность:
Время: O(n)
Память: O(1)
'''
def linear_search(arr, target):
    #Проходим по списку по очереди
    for i in range(len(arr)):

        #Если нашли нужное число
        if arr[i] == target:
            return i  #Bозвращаем индекс

    #Если число не нашли
    return -1

#Проверка
print(linear_search([10, 50, 30, 70, 80, 20], 30))
```
